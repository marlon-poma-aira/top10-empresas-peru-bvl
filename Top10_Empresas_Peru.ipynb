```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df_empresas = pd.read_csv('data/Top10_Empresas_Peru_BVL.csv')

df_empresas = df_empresas.sort_values('IngresosFY_MilesMillonesPEN', ascending= True)

color_sector = {
    'Finanzas': '#1f77b4',
    'Comercio minorista' : '#ff7f0e',
    'Mineria' : '#8c564b',
    'Consumo masivo' : '#2ca02c',
    'Servicios de distribucion' : '#7f7f7f',
}

colores = df_empresas['Sector'].map(color_sector)

plt.figure(figsize=(11,6))
plt.barh(df_empresas['Empresa'], df_empresas['IngresosFY_MilesMillonesPEN'], color=colores)
plt.title('Top 10 empresas peruanas que cotizan en bolsa -Ingresos anuales (BVL)', fontsize=12, fontweight='bold')
plt.xlabel('Ingresos (miles de millones de PEN)')

# Leyenda manual por sector
from matplotlib.patches import Patch
handles = [Patch(color=c, label=s) for s, c in color_sector.items()]
plt.legend(handles=handles, loc='lower right', fontsize=8)

plt.grid(alpha= 0.3, axis='x')
plt.tight_layout()
plt.savefig('grafico_top10_ingresos.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
df_bpa = df_empresas.sort_values("CrecimientoBPA_TTM_YoY_Pct", ascending=True)
colores_bpa = ["#2ca02c" if x > 0 else "#d62728" for x in df_bpa["CrecimientoBPA_TTM_YoY_Pct"]]

plt.figure(figsize=(11, 6))
plt.barh(df_bpa["Empresa"], df_bpa["CrecimientoBPA_TTM_YoY_Pct"], color=colores_bpa)
plt.title("Crecimiento de utilidad por acción (BPA) — Variación % interanual", fontsize=12, fontweight="bold")
plt.xlabel("Crecimiento % BPA (TTM YoY)")
plt.axvline(0, color="black", linewidth=0.8)
plt.grid(alpha=0.3, axis="x")
plt.tight_layout()
plt.savefig("grafico_top10_crecimiento_bpa.png", dpi=300, bbox_inches="tight")
plt.show()
```
